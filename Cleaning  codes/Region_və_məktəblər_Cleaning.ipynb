```python
import pandas as pd
import unicodedata
from clean_region_names import replacing
```

```python
region_school_df = pd.read_csv("Region_və_məktəblərin_siyahısı.csv")
```

```python
# ş. və ray. əvəz edək. funksiya şəkilndə istifadə edirik çünki digər dataları təmizləyərkən eyni koda ehtiyyac duyulacaq

def replacing(df, col_name):

    df[f"{col_name}"] = (
        df[f"{col_name}"]
        .str.lower()
        .str.replace("baki ş.", "bakı şəhəri", regex=False)
        .str.replace("ray.", "rayonu", regex=False)
        .str.replace("bakı ş.", "bakı şəhəri", regex=False)
        .str.strip()
    )

    rules = {
        "yasamal": "bakı şəhəri yasamal rayonu",
        "binəqədi": "bakı şəhəri binəqədi rayonu",
        "xətai": "bakı şəhəri xətai rayonu",
        "nərimanov": "bakı şəhəri nərimanov rayonu",
        "qaradağ": "bakı şəhəri qaradağ rayonu",
        "naxçıvan şəhəri": "naxçıvan şəhəri",
        "şəmkir": "şəmkir rayonu",
        "yevlax": "yevlax rayonu",
        "şəki": "şəki rayonu",
        "qubadli": "qubadlı rayonu",
        "laçin": "laçın rayonu",
        "lənkəran": "lənkəran rayonu",
        "masalli": "masallı rayonu",
        "cəbrayil": "cəbrayıl rayonu",
        "xanlar": "göygöl rayonu",
        "dəvəçi": "şabran rayonu",
        "səbayel": "bakı şəhəri səbail rayonu",
        "türk liseyləri": "xocavənd rayonu",
        "əli-bayramlı": "şirvan şəhəri",
        "şuşa": "şuşa şəhəri",
        "naxçivan şəhəri̇": "naxçıvan şəhəri",
        "naxçivan mr culfa rayonu": "naxçıvan mr culfa rayonu"
    }

    def replace_region(cell):
        if pd.isna(cell):
            return cell

        text = cell.lower()

        for key, value in rules.items():
            if key in text:
                return value.title()

        return cell

    df[col_name] = df[col_name].apply(replace_region)

 # Bölgə sütu   
    def normalize_text(s):
        if pd.isna(s):
            return s
        s = unicodedata.normalize("NFKD", s)
        s = "".join(c for c in s if not unicodedata.combining(c))
        return s.lower().strip()

    df["Bölgə_norm"] = df[col_name}].apply(normalize_text)

    for key, value in rules.items():
        key_norm = normalize_text(key)
        mask = region_school_df["Bölgə_norm"].str.contains(key_norm, na=False)
        df.loc[mask, col_name] = value

# Bütün sətrlər kiçik ilə yazılsın
    df[col_name] = (
        df[col_name]
        .str.lower())


    
```

```python
replacing(region_school_df, "Bölgə")
```
