```python
#Bu sütunlar üzrə dəyərlər rəsmi mənbədə hesablanmadığı (NULL olduğu) üçün analitik nəticələrin təhrif olunmaması məqsədilə analizdən çıxarılmışdır.
df_intervals = df_intervals.dropna(
    subset=['Şagirdlərin nisbi sayı', 'Reytinq']
)
```

```python
# Bölgə sütunundakı sətrləri replacing funksiyası ilə təmizləyirik
replacing(df_intervals, "Bölgə")
```

```python
# Reytinq və Şagirdlərin nisbi sayına uyğun olaraq hər fənn üzrə həmin ildəki respublika ortalama nisbi sayını tapırıq.
df_intervals["Ortalama nisbi say"] = round(df_intervals["Şagirdlərin nisbi sayı"]/ df_intervals["Reytinq"], 2)
```

```python
df_intervals[(df_intervals["İl"]==2019) & (df_intervals["Fənn"] == "Ana dili") 
                                        & (df_intervals["Buraxılış_imtahanları"] == "tam orta təhsil səviyyəsi")].head(10)
```

```python
# Şagirdlərin nisbi sayı	və Ortalama nisbi say sütunları üçün iki dəfə run edirəm.

group_cols = ['Bölgə', 'Məktəb', 'Fənn', 'İl', 'Buraxılış_imtahanları']

df = df_intervals.copy()

# ===============================
# 1. interval_min çıxarılır
# ===============================
df['interval_min'] = (
    df['Ümumi balların intervalları']
    .astype(str)
    .str.extract(r'^(\d+)')
    .astype(float)
)

# yalnız lazım olan intervallar
needed = [20, 40, 60, 80]
df_needed = df[df['interval_min'].isin(needed)]

# ===============================
# 2. Pivot (sürətin açarı)
# ===============================
pivot = (
    df_needed
    .pivot_table(
        index=group_cols,
        columns='interval_min',
        values='Ortalama nisbi say',
        aggfunc='first'
    )
)

# ===============================
# 3. Yeni intervallar (vectorized)
# ===============================
pivot['20–40'] = pivot[20] - pivot[40]
pivot['40–60'] = pivot[40] - pivot[60]
pivot['60–80'] = pivot[60] - pivot[80]
pivot['80–100_new'] = pivot[80]

# ===============================
# 4. İki yeni sütunu əsas df-ə qaytar
# ===============================
new_cols = (
    pivot[['20–40', '40–60', '60–80', '80–100_new']]
    .reset_index()
)

df_intervals = df.merge(
    new_cols,
    on=group_cols,
    how='left'
)
```

```python
# Şərtlərə uyğun sıfırlanacaq sütunlar
reset_rules = {
    "20 – 100 (20 daxil olmamaq şərtilə)": ['60–80', '40–60', '80–100_new'],
    "40 – 100 (40 daxil olmamaq şərtilə)": ['20–40', '60–80', '80–100_new'],
    "60 – 100 (60 daxil olmamaq şərtilə)": ['20–40', '40–60', '80–100_new'],
    "80 – 100 (80 daxil olmamaq şərtilə)": ['20–40', '40–60', '60–80'],
    # istəsəniz daha çox şərt əlavə edə bilərsiniz
}

# loop ilə tətbiq
for condition, cols_to_zero in reset_rules.items():
    df_intervals.loc[
        df_intervals['Ümumi balların intervalları'] == condition,
        cols_to_zero
    ] = 0


```

```python
# mapping dictionary
interval_map = {
    "80 – 100 (80 daxil olmamaq şərtilə)": "80 - 100",
    "60 – 100 (60 daxil olmamaq şərtilə)": "60 - 80",
    "40 – 100 (40 daxil olmamaq şərtilə)": "40 - 60",
    "20 – 100 (20 daxil olmamaq şərtilə)": "20 - 40"
}

# yeni sütunu əlavə edirik
df_intervals['Balların intervalları'] = df_intervals['Ümumi balların intervalları'].map(interval_map)
```

```python
# Yeni intervallara uyğun olaraq Şagirdlərin nisbi sayları
df_intervals["Şagirdlərin nisbi sayı (faizlə)"] = df_intervals["20–40"] + df_intervals["40–60"] + df_intervals["60–80"] + df_intervals["80–100_new"]
```

```python
# Yeni intervallara uyğun olaraq Şagirdlərin respublika üzrə ortalama nisbi sayları
df_intervals["Şagirdlərin ortalama nisbi sayı (faizlə)"] = df_intervals["20–40"] + df_intervals["40–60"] + df_intervals["60–80"] + df_intervals["80–100_new"]
```

```python
# Hesablama zaman; yaratdığımız amma artıq istifadə olunmayan sütunları silinir
df_intervals.drop(columns=['interval_min', '20–40','40–60','60–80','80–100_new'], inplace=True)
```

```python
df_intervals[(df_intervals["İl"]==2019) & (df_intervals["Fənn"] == "Ana dili") 
                                        & (df_intervals["Buraxılış_imtahanları"] == "tam orta təhsil səviyyəsi")].head()
```

```python
# Yeni intervallar və dəyərlərə uyğun olaraq Şagirdlərin reytinqi
df_intervals["Şagirdlərin reytinqi (yeni)"] = round(df_intervals["Şagirdlərin nisbi sayı (faizlə)"] / df_intervals["Şagirdlərin ortalama nisbi sayı (faizlə)"],3)
```
