```python
import pandas as pd
from clean_region_names import replacing
```

```python
df_highSchool_enternace = pd.read_csv("Ali_məktəblərə_qəbulolma_göstəricilər.csv")
```

```python
df_highSchool_enternace.head()
```

```python
df_highSchool_enternace.isna().sum()
```

```python
df_highSchool_enternace.drop_duplicates(inplace=True)
```

```python
df_highSchool_enternace['İmtahanda iştirak edənlərin sayı'] = df_highSchool_enternace['İmtahanda iştirak edənlərin sayı'].astype(int)
df_highSchool_enternace['İmtahanda iştirak edənlərin orta balı'] = df_highSchool_enternace['İmtahanda iştirak edənlərin orta balı'].astype(float)
```

```python
# () olan sətrlər var, () daxilindəki dəyəri silirik, beləcə number tipinə çevirmdə error yaşanmır.
cols = ['İmtahanda iştirak edənlərin sayı', 'İmtahanda iştirak edənlərin orta balı']

for col in cols:
    df_highSchool_enternace[col] = (
        df_highSchool_enternace[col]
        .astype(str)                      # hər şeyi string et
        .str.replace(r'\(.*?\)', '', regex=True)  # () və içindəkiləri sil
        .str.strip()                      # boşluqları təmizlə
        .replace('---', '0')              # --- varsa 0 et
    )
```

```python
df_highSchool_enternace['İmtahanda iştirak edənlərin sayı'] = df_highSchool_enternace['İmtahanda iştirak edənlərin sayı'].astype(int)
df_highSchool_enternace['İmtahanda iştirak edənlərin orta balı'] = df_highSchool_enternace['İmtahanda iştirak edənlərin orta balı'].astype(float)
```

```python
df_highSchool_enternace.info()
```

```python
df_highSchool_enternace.to_csv("Ali_məktəblərə_qəbulolma_göstəricilər_cleaned.csv", index=False, encoding="utf-8-sig")
```
