```python

import time
import requests
import pandas as pd
from bs4 import BeautifulSoup
from collections import Counter
from urllib3.util.retry import Retry
from requests.adapters import HTTPAdapter

```

```python
# Bütün verilən bölgələrin mövcud olduğu səhifələrin nömrələrini tapmaq
regions_nums = []

for value in range(10, 100):

    url = f"http://mekteb.edu.az/az/smenu{value}.htm"
    response = requests.get(url, timeout=10)

    if response.status_code == 200:
        regions_nums.append(value)

print(regions_nums)
print(len(regions_nums))
```

```python
# Hər bir bölgə üçün mövcud olan məktəblərin səhifələrinin linklərini tapmaq
regions_links = []

for regions_num in regions_nums:
    for value in range(0, 999):

        value_str = f"{value:03d}"  # 0–9 -> 000, 1 -> 001, 10 -> 010, 99 -> 099

        url = f"http://mekteb.edu.az/az/{regions_num}/{regions_num}{value_str}a0000_2a.htm"
        response = requests.get(url, timeout=10)

        if response.status_code == 200:
            regions_links.append(url)

print(regions_links)
print(len(regions_links))
```

```python
# Yardımçı cleaning funksiyaları

# ===============================================
# Numeric dəyərlərdə yer alan mətnin təmizlənməsi
def clean(x):
    if x is None:
        return None
    x = x.strip()
    return None if x == "---" else x
# ===============================================

# ===============================================
# IndexError-ların qarşısını almaq üçün təhlükəsiz text funksiyası
def safe_text(tds, idx):
    return clean(tds[idx].get_text(strip=True)) if idx < len(tds) else None
# ===============================================
```

```python
# Requests kitabxanası ilə HTTP sorğuları göndərmək üçün

#============================================================
session = requests.Session() # eyni serverə təkrar sorğular üçün

retries = Retry(
    total=5,
    backoff_factor=1.5,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)

adapter = HTTPAdapter(max_retries=retries)
session.mount("http://", adapter)

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"  # bot detection-dan qaçmaq üçün
}

rows = [] # Ali məktəblərə qəbulolma göstəriciləri üçün siyahı
rows_subjects = [] # mtahan fənləri üzrə cavabların göstəriciləri üçün siyahı
#============================================================

#============================================================
# Bütün məktəblərin əsas səhifələrinin URL-lərinin yüklənməsi
for url in all_school_urls:

    try:
        page = session.get(url, headers=HEADERS, timeout=(5, 20))
    except Exception as e:
        print("Əsas səhifə timeout:", e)
        continue

    soup = BeautifulSoup(page.text, "html")
#============================================================

#============================================================
    # Məktəb və region adlarının çıxarılması
    trs = [tr for tr in soup.find_all("tr") if tr.find("td")] # boş sətrləri at
    info_tr = trs[1]

    school_name = info_tr.find_all("td")[2].get_text(strip=True)
    region_raw = info_tr.find_all("td")[4].get_text(strip=True)
    region_name = region_raw.split(":")[-1].strip().title()
#============================================================

    years = [opt.get_text(strip=True) for opt in soup.find_all("option")] # mövcud illər

#============================================================
    # Hər il üçün ayrıca səhifəyə sorğu göndərilməsi və məlumatların çıxarılması
    for year in years:

        base, _ = url.split("_", 1)
        year_url = base[:-4] + year + "_1a.htm" # scrapping həyata keçirmək istədiyimiz səhifənin URL-i

        try:
            r = session.get(year_url, headers=HEADERS, timeout=(5, 20))
        except Exception:
            continue

        if r.status_code != 200:
            continue

        year_soup = BeautifulSoup(r.text, "html")

        head = year_soup.find("span", class_="head")
        if not head or "kampaniya" not in head.get_text(strip=True).lower(): # səhifənin düzgün yüklənməməsi halında head içində "kampaniya" sözü yoxdursa
            continue

        tables = year_soup.find_all(
            "table",
            {"border": "0", "cellspacing": "1", "bgcolor": "#85B0C7", "cellpadding": "2"}
        )

        subjects_table = tables[0]  # Fənlər üzrə nəticələr cədvəli (tables[0])

        for tr in subjects_table.select("tbody tr"):
            tds = tr.find_all("td")

            
            if len(tds) < 10: # Header və natamam sətrləri at
                continue

            subject = tds[0].get_text(strip=True)

            if subject.strip().lower() == "cəmi": # "Cəmi" sətrini istisna et (analitik olaraq lazımsızdır)
                continue

            data = {
                "Oğlan": {
                    "count": safe_text(tds, 1),
                    "percent": safe_text(tds, 4),
                    "rating": safe_text(tds, 7),
                },
                "Qız": {
                    "count": safe_text(tds, 2),
                    "percent": safe_text(tds, 5),
                    "rating": safe_text(tds, 8),
                }
            }

            for gender, v in data.items():

               
                if not v["count"]:  # boş və "---" sətrləri xammalda saxlamırıq
                    continue

                rows_subjects.append({
                    "İl": int(year),
                    "Bölgə": region_name,
                    "Məktəb": school_name,
                    "Fənn": subject,
                    "Cins": gender,
                    "İmtahanda iştirak edənlərin sayı": v["count"],
                    "Düzgün cavabların faizi": v["percent"],
                    "Reytinq": v["rating"],
                })

        if len(tables) < 2: # Cədvəl mövcud deyilsə
            continue

        table = tables[1]
        current_group = None


        for tr in table.select("tbody tr"):
            tds = tr.find_all("td")
            if not tds:
                continue

            if "tablestat1" in tds[0].get("class", []):
                current_group = tds[0].get_text(strip=True)
                
                if current_group.strip().lower() == "cəmi":
                    current_group = "__SKIP__"
                    continue

                data_tds = tds[1:]
            else:
                if current_group == "__SKIP__":
                    continue

                data_tds = tds

            gender_raw = data_tds[0].get_text().strip().lower()
            if gender_raw == "*":
                continue
            elif gender_raw == "o":
                gender = "Oğlan"
            elif gender_raw == "q":
                gender = "Qız"
            else:
                continue

            exam_count     = safe_text(data_tds, 1)
            exam_avg       = safe_text(data_tds, 2)
            accepted_total = safe_text(data_tds, 3)
            accepted_avg   = safe_text(data_tds, 5)

            state_free = int(safe_text(data_tds, 6) or 0)
            state_paid = int(safe_text(data_tds, 7) or 0)
            private    = int(safe_text(data_tds, 8) or 0)

            base_row = {
                "İl": int(year),
                "Bölgə": region_name,
                "Məktəb": school_name,
                "İxtisas qrupu": current_group,
                "Cins": gender,
                "İmtahanda iştirak edənlərin sayı": exam_count,
                "İmtahanda iştirak edənlərin orta balı": exam_avg,
                "Qəbul olanların sayı": accepted_total,
                "Qəbul olanların orta balı": accepted_avg,
            }

# əvvəlki base_row və append hissəsi dəyişir

            for inst, pay_type, count in [
                ("dövlət", "ödənişsiz", state_free),
                ("dövlət", "ödənişli", state_paid),
                ("özəl", "özəl", private)
            ]:
                if count and count != 0:  # SAY 0-dan fərqli olmalıdır
                    rows.append({**base_row, "Təhsil müəssisəsi": inst, "Təhsil əsası": pay_type, "Sayı": count})


        time.sleep(0.7)  # serveri yormamaq üçün

df = pd.DataFrame(rows)
df_subjects = pd.DataFrame(rows_subjects)
```

```python
# 1. Nəticələri CSV faylına yaz
csv_file = "İmtahan_fənləri_üzrə_cavabların_göstəricilər.csv"

df_subjects.to_csv(
    csv_file,
    index=False,
    encoding="utf-8-sig"
)
```

```python
# Requests kitabxanası ilə HTTP sorğuları göndərmək üçün
session = requests.Session()
retries = Retry(
    total=5,
    backoff_factor=1.5,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)
adapter = HTTPAdapter(max_retries=retries)
session.mount("http://", adapter)

HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
#=======================================================
records_gender = [] # Fənlər üzrə nisbi balların dinamikası üçün siyahı
records_interval = [] # Şagirdlərin nisbi saylarının müvəffəqiyyət kateqoriyaları üçün siyahı
for url in all_school_urls:

    new_urls = [url.split("_")[0] + "_7a.htm", url.split("_")[0] + "_6a.htm"]

    for new_url in new_urls:

        buraxılış_imtahanları = (
            "tam orta təhsil səviyyəsi" if "_7a.htm" in new_url
            else "ümumi orta təhsil səviyyəsi"
        )

        try:
            page = session.get(new_url, headers=HEADERS, timeout=(5, 20))
            page.raise_for_status()
        except Exception as e:
            print("Səhifə yüklənmədi:", new_url, e)
            continue

        page.encoding = "windows-1254"
        soup = BeautifulSoup(page.text, "html.parser")

        trs = [tr for tr in soup.find_all("tr") if tr.find_all("td")]

        if len(trs) < 2:
            print("Info sətri tapılmadı:", new_url)
            continue

        info_tds = trs[1].find_all("td")

        if len(info_tds) < 5:
            print("Info td-lər çatışmır:", new_url)
            continue

        school_name = info_tds[2].get_text(strip=True)
        region_name = info_tds[4].get_text(strip=True).split(":")[-1].strip().title()

        # -------- FƏNNLƏRİ AVTOMATİK TAP --------
        b_tags = soup.find_all("b")
        subjects_raw = [b.get_text(strip=True) for b in b_tags if b.get_text(strip=True)]
        counts = Counter(subjects_raw)
        subjects = [s for s in subjects_raw if counts[s] > 1]


        # -------- table html taglarini AVTOMATİK TAP --------
        separator_node = None

        for b_tag in soup.find_all("b"):
            b_text = " ".join(b_tag.stripped_strings).lower()
            if "b) şagirdlərin" in b_text:
                separator_node = b_tag
                break

        if separator_node is None:
            raise ValueError("Ayırıcı <b> tag-ı tapılmadı")

        tables_above = [
            t for t in separator_node.find_all_previous("table")
            if t.get("bgcolor") == "#85B0C7"
        ][::-1]

        tables_below = [
            t for t in separator_node.find_all_next("table")
            if t.get("bgcolor") == "#85B0C7"
        ]
        # -------- fənləri html taglarini AVTOMATİK TAP --------
        separator_node = None
        for b in soup.find_all("b"):
            b_text = " ".join(b.stripped_strings).lower()
            if "b) şagirdlərin" in b_text:
                separator_node = b
                break

        if separator_node is None:
            raise ValueError("Ayırıcı <b> tag-ı tapılmadı")

        # 2) ayırıcıdan YUXARIDAKI fənn adları (<b>)
        subjects_above = [
            b.get_text(strip=True)
            for b in separator_node.find_all_previous("b")
            if b.get_text(strip=True)
        ][::-1]

        # a) bölmə başlığını çıxar
        if subjects_above:
            subjects_above = subjects_above[1:]

        # 3) ayırıcıdan AŞAĞIDAKI fənn adları (<b>)
        subjects_below = [
            b.get_text(strip=True)
            for b in separator_node.find_all_next("b")
            if b.get_text(strip=True)
        ]

        # -------- FIRST HALF (GENDER) --------
        for table_index, table in enumerate(tables_above):

            subject = subjects_above[table_index] if table_index < len(subjects_above) else None
            rows = table.find_all("tr")

            if len(rows) < 3:
                continue

            years = [int(td.get_text(strip=True)) for td in rows[1].find_all("td")]

            for row in rows[2:]:
                tds = row.find_all("td")
                if len(tds) < 2:
                    continue

                gender = tds[0].get_text(strip=True)

                for i, cell in enumerate(tds[1:]):
                    text = cell.get_text("\n", strip=True)
                    if "\n" not in text or i >= len(years):
                        continue

                    raw_faiz, raw_reytinq = text.split("\n")
                    faiz = clean(raw_faiz)
                    reytinq = clean(raw_reytinq)

                    records_gender.append({
                        "İl": years[i],
                        "Bölgə": region_name,
                        "Məktəb": school_name,
                        "Fənn": subject,
                        "Cins": gender,
                        "Düzgün cavabların faizi": float(faiz) if faiz else None,
                        "Reytinq": float(reytinq) if reytinq else None,
                        "Buraxılış_imtahanları": buraxılış_imtahanları
                    })

        # -------- SECOND HALF (INTERVAL) --------
        for table_index, table in enumerate(tables_below):

            subject = subjects_below[table_index] if table_index < len(subjects_below) else None
            rows = table.find_all("tr")

            if len(rows) < 3:
                continue

            years = [int(td.get_text(strip=True)) for td in rows[1].find_all("td")]

            for row in rows[2:]:
                tds = row.find_all("td")
                if len(tds) < 2:
                    continue

                interval = tds[0].get_text(" ", strip=True)

                for i, cell in enumerate(tds[1:]):
                    text = cell.get_text("\n", strip=True)
                    if "\n" not in text or i >= len(years):
                        continue

                    raw_nisbi, raw_reytinq = text.split("\n")
                    nisbi = clean(raw_nisbi)
                    reytinq = clean(raw_reytinq)

                    records_interval.append({
                        "Bölgə": region_name,
                        "Məktəb": school_name,
                        "Fənn": subject,
                        "İl": years[i],
                        "Ümumi balların intervalları": interval,
                        "Şagirdlərin nisbi sayı": float(nisbi) if nisbi else None,
                        "Reytinq": float(reytinq) if reytinq else None,
                        "Buraxılış_imtahanları": buraxılış_imtahanları
                    })

df_gender = pd.DataFrame(records_gender)
df_interval = pd.DataFrame(records_interval)
```

```python
session = requests.Session()
retries = Retry(
    total=5,
    backoff_factor=1.5,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)
adapter = HTTPAdapter(max_retries=retries)
session.mount("http://", adapter)

HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}

records = []

for new_url in all_school_urls:

    try:
        page = session.get(new_url, headers=HEADERS, timeout=(5, 20))
        page.raise_for_status()
    except Exception as e:
        print("Səhifə yüklənmədi:", new_url, e)
        continue

    page.encoding = "windows-1254"
    soup = BeautifulSoup(page.text, "html.parser")

    # ⬅️ INFO SƏTRİNİ HƏR SƏHİFƏ ÜÇÜN YENİDƏN TAP
    trs = [tr for tr in soup.find_all("tr") if tr.find("td")]

    if len(trs) < 2:
        continue

    info_tds = trs[1].find_all("td")

    if len(info_tds) < 5:
        continue

    school_name = info_tds[2].get_text(strip=True)
    region_name = info_tds[4].get_text(strip=True).split(":")[-1].strip().title()

    records.append({
        "Bölgə": region_name,
        "Məktəb": school_name
    })

df_region_and_schools = pd.DataFrame(records)

    
```

```python
csv_file = "Region_və_məktəblərin_siyahısı.csv"

df_region_and_schools.to_csv(
    csv_file,
    index=False,
    encoding="utf-8-sig"
)
```
